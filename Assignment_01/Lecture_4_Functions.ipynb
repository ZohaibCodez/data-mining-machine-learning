```python
# find the purpose of math.copysign(x,y) function and run it with sample data
# math.copysign(x, y) is used to take the magnitude (absolute value) of x and give it the sign of y
import math

print(math.copysign(10, 5))
print(math.copysign(10, -5))
print(math.copysign(-10, 5))
print(math.copysign(-10, -5))
```

```python
# find the purpose of math.trunc(x) function and run it with sample data
# math.trunc(x) removes the decimal part of a number and returns the integer part.
print(math.trunc(10.75))
print(math.trunc(10.25))
print(math.trunc(-10.75))
print(math.trunc(-10.25))
```

```python
# find the purpose of math.fabs(x) function and run it with sample data
# math.fabs(x) returns the absolute value of a number.
print(math.fabs(10))
print(math.fabs(-10))
print(math.fabs(10.75))
print(math.fabs(-10.75))
```

# 4. Adding new functions

```python
def print_lyrics():
    print ("I'm a lumberjack, and I'm okay.")
    print ('I sleep all night and I work all day.')
```

```python
print (print_lyrics)
type(print_lyrics)
```

```python
print_lyrics()
```

```python
def repeat_lyrics():
    print_lyrics()
    print_lyrics()
```

```python
repeat_lyrics()
```

```python
# parameterise function
def print_twice(bruce):
    print (bruce)
    print (bruce)
```

```python
print_twice('hi')
```

```python
print_twice(3)
```

```python
print_twice("pak" * 3) # The argument is evaluated first, before the function is called
```

```python
# To return a result from a function, we use the return statement in our function.
def add(a, b):
    added = a + b
    return added
```

```python
print (add(3,4))
```

```python
# define a function, to which when your name is passed as an argument, returns the no of characters in your name
def count_chars(name:str)->int:
    return len(name)

print(count_chars("Zohaib"))
```

```python
# define a function to find the maximum of three numbers
def find_max(a, b, c):
    if a > b and a > c:
        return a
    elif b > a and b > c:
        return b
    else:
        return c

find_max(3,2,1)
```

```python
# define a function that accepts an argument which is a string and you have to reverse that string
def reverse_string(text:str):
  reversed:str = ""
  for char in text:
    reversed = char + reversed
  return reversed

reverse_string("Zohaib")
```
